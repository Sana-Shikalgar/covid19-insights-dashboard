## Calculate and store the missing values and % of missing values


```python
# Count missing values
missing_counts = df.isnull().sum()

# Calculate percentage of missing data
missing_percentage = (missing_counts / len(df)) * 100

# Convert to DataFrame for nicer display
missing_df = missing_counts.reset_index()
missing_df.columns = ['Column', 'Missing Values']

# Add percentage column
missing_df['% Missing'] = missing_percentage.values

# Sort by descending missing values
missing_df = missing_df.sort_values(by='Missing Values', ascending=False)

# Display the missing data information
print(missing_df)
```

```python
# Save missing data summary to Excel (editable in Excel)
missing_df.to_csv('data_quality_analysis.csv', index=False)

print("✅ Files saved: data_quality_analysis.csv")
```

## Extract Selected Columns from COVID-19 Dataset

Manually selected the feilds to keep and remove as it would help with how to analyse the remaining data. The new excel is named: _'feild_decision.xlsx'_ in folder _'decision_on_feilds'_

I have also stated the reason for each feild to be kept _(KEEP)_ or removed _(REMOVE)_.


### Get the columns that are selected


```python
# Path to the decision file
PARENT_DIR = os.path.dirname(os.getcwd())  # Do not use BASE_DIR here, it gives error 
DECISION_FILE_PATH = os.path.join("feild_decision.xlsx")

filter_df = pd.read_excel(DECISION_FILE_PATH) # Read the Excel file

# Filter rows where Decision is KEEP
keep_df = filter_df[filter_df['Decision'].str.upper() == 'KEEP']

# Select only 'Column' which has Column names to keep
keep_cols = keep_df['Column'].to_list()

# Print the result
print(keep_cols)
```

### Create a new dataset containing only the selected columns from the original data


```python
# Output file path
output_file = "filtered_covid_data.csv"

# Select only the columns to keep (ignore columns not present)
columns_to_extract = [col for col in keep_cols if col in df.columns]
new_df = df[columns_to_extract]

# print("Filtered Data Shape:", new_df.shape) # Returns (429435, 11)

# Save the filtered dataset
new_df.to_csv(output_file, index=False)

print(f"Filtered dataset saved as {output_file} with {len(new_df.columns)} columns.")
```
